```python
import numpy as np
import pandas as pd

prices = np.array([2.95, 4.95, 1.95, 12.95])
quantities = np.array([3, 1, 0, 2])
```

```python
print(prices)
```

```python
total_price = prices*quantities
print(total_price)
```

```python
orders = np.array([
    [3, 0, 0, 2],
    [10, 3, 4, 8],
    [19, 0, 1, 1]
 ])
```

```python
sum_total = total_price.sum()
sum_total
```

```python
values = orders*prices
print(values)
```

```python
prod_A = values[:,0]
print(prod_A)
```

```python
prod_A_value = prod_A.sum()
```

```python
print(f'Value sold by product A is {prod_A_value} euros')
```

```python
prod_B = values[:,1]
prod_B_value = prod_B.sum()
print(f'Value sold by product B is {prod_B_value:.2f} euros')
```

```python
prod_C = values[:,2]
prod_C_value = prod_C.sum()
print(f'Value sold by product C is {prod_C_value:.2f} euros')
```

```python
prod_D = values[:,3]
prod_D_value = prod_D.sum()
print(f'Value sold by product D is {prod_D_value:.2f} euros')
```
